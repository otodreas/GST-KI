# GST analysis
## Load & summarize data

The gnomAD data is summarized by Ensembl canonical transcript coverage of variants using the same methodology as Gonzalez-Padilla.

```python
import os
from pathlib import Path
import polars as pl
import otodreas_gst_ki.gnomad.load as load

datadirs = [Path().resolve().parent / "data/gnomad/2026_11_12", Path().resolve().parent / "data/gnomad/2026_11_17"]
filepaths = []

for d in datadirs:
    filenames = os.listdir(d)
    filepaths.extend([d / f for f in filenames])

datasets_dict = {}

for f in filepaths:
    data = load.load_gnomad(f)
    gene = data.gene
    data.summarize()
    datasets_dict.update({gene: data})
```

```python
# Check if there are any shared variants across genes
shared_variants_across_genes = load.find_shared_variants(list(datasets_dict.values()))

if len(shared_variants_across_genes) > 0:
    print(shared_variants_across_genes.sort("nGenes", descending=True))
```

```python
# Concatenate all datasets
full_data = load.concat_gnomad(list(datasets_dict.values()))
full_data.head()
```

### Fig 1D

Fig 1D only includes exonic variants, so this is not exactly aligned with Gonzalez-Padilla either, since I have not selected exonic regions as rigorously as her yet. However, the binning of variants according to MAF is the same as in her paper.

```python
df_1d = (
    full_data
    .filter(
        pl.col("Exonic") & pl.col("CanonicalTranscript")
    )
    .select("Allele Frequency")
    .rename({"Allele Frequency": "MAF of exonic variants"})
)

def get_bin_pct(
    bounds: tuple[float, float],
    df: pl.DataFrame = df_1d,
    var: str = "MAF of exonic variants"
) -> float:
    lo, hi = bounds
    n = df.filter((pl.col(var) > lo) & (pl.col(var) <= hi)).height
    return 100 * n / df.height

bins = {
    "Singleton (<= 1e-5)": (0, 1e-5),
    "Very rare (1e-5 - 1e-3)": (1e-5, 1e-3),
    "Rare (1e-3 - 1e-2)": (1e-3, 1e-2),
    "Common (> 1e-2)": (1e-2, 1),
}

print(f"Percent of {len(df_1d)} exonic variants in canonical transcripts whose MAF is:")
for label, b in bins.items():
    print(f"  {label}: {get_bin_pct(b):.1f}%")

fig, ax = plt.subplots()

sns.histplot(
    data=df_1d.to_pandas(),
    x="MAF of exonic variants",
    cumulative=True,
    log_scale=True,
    element="step",
    fill=False,
    stat="percent",
    ax=ax
)

ax.set_ylabel("Percentage of variants")

plt.show()
```

### Fig 1E

We determined that the expected and observed SNV matrix for synonymous and missense variants would suffice for reproducing Gonzalez-Padilla's fig 1E, given that the observed synonymous and missense SNVs were close to the actual number of missense and synonymous VEP annotations in the gnomAD CSVs. The observed SNV numbers are reported following a filter in gnomAD

![filter](assets/observed_snvs_filter.png)

All the filters are reproducible with data from the CSVs with the exception of the depth filter, since depth is not reported in the CSVs. Therefore, we expect the number of missense and synonymous SNVs to be the same or slightly larger in the reproduced set compared to the observed field in the gnomAD browser. Results below show that assumption is correct.

```python
df_metadata = pl.read_csv(Path().resolve().parent / "data/gnomad/metadata/custom_metadata.csv")

df_metadata = df_metadata.with_columns(
    (pl.col("exp_mis") / pl.col("exp_syn")).alias("exp_missyn"),
    (pl.col("obs_mis") / pl.col("obs_syn")).alias("obs_missyn")
)


cat_recomputed = {
    "missense_variant": "mis. recomputed",
    "synonymous_variant": "syn. recomputed",
}
cat_overview = {
    "obs_mis": "mis. from overview",
    "obs_syn": "syn. from overview",
}

frames = []
for gene, ds in datasets_dict.items():
    recomputed = (
        ds.dataMatrix
        .filter(
            pl.col("VEP Annotation").str.contains_any(["missense_variant", "synonymous_variant"]),
            pl.col("CanonicalTranscript"),
            pl.col("Allele Frequency") < 0.1,
        )
        .group_by("VEP Annotation")
        .len()
        .select(
            pl.lit(gene).alias("gene"),
            pl.col("len").cast(pl.Int64),
            pl.col("VEP Annotation").replace_strict(cat_recomputed).alias("category"),
        )
    )

    overview = (
        df_metadata
        .filter(pl.col("gene") == gene)
        .select(pl.col("gene"), "obs_mis", "obs_syn")
        .unpivot(index="gene", on=["obs_mis", "obs_syn"],
                 variable_name="category", value_name="len")
        .select(
            "gene",
            pl.col("len").cast(pl.Int64),
            pl.col("category").replace_strict(cat_overview),
        )
    )

    frames += [recomputed, overview]

combined = pl.concat(frames).sort("gene", "category")

fig, ax = plt.subplots()

sns.barplot(
    combined.to_pandas(),
    x="len",
    y="gene",
    hue="category"
)

ax.legend(
    title="metric",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    borderaxespad=0,
)

ax.set_xlabel("Number of variants")

plt.title("recomputed observed snv count is close to and never smaller than the count reported on the gene summary on gnomad")
```

## Key takeaways

- There are a similar number of exonic and intronic variants across the variants, unlike UGT
- Unlike Gonzalez-Padilla's work on UGT, most of the variants do not fall into the classes $\{ \text{Stop gained, frameshift, synonymous, missense}\}$
- Of the classes mentioned above, missense appears the most frequently across genes, followed by synonymous. The same pattern was observed by Gonzalez-Padilla
- MAF of exonic variants follows a similar distribution as it did for UGT, with similar number of variants in each bin (the same bins as Gonzalez-Padilla used)
- Observed missense & synonymous SNV counts from the constraint table in the gnomAD browser are nearly reproducible from the variant-level data presented in the browser
- There is a positive relationship between expected missense/synonymous ratio and observed missense/synonymous ratio for GST genes. The same was true for UGT
